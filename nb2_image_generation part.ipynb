```python
import torch
import gc
import re
import json
import os
from fastapi import FastAPI
from fastapi.responses import FileResponse
from pydantic import BaseModel
from typing import List, Optional
import nest_asyncio
import uvicorn
from pyngrok import ngrok
from diffusers import StableDiffusionXLPipeline, DPMSolverMultistepScheduler
from compel import Compel, ReturnedEmbeddingsType
from IPython.display import display

# ==========================================
# 0. PREVENT OOM CRASHES ON RERUNS
# ==========================================
if 'pipe' in globals():
    del pipe
if 'compel' in globals():
    del compel
if 'image' in globals():
    del image
gc.collect()
torch.cuda.empty_cache()

# ==========================================
# 1. INITIALIZE PIPELINE ONCE (Global Setup)
# ==========================================
print("🔄 Loading RealVisXL V4.0 into GPU...")
model_id = "SG161222/RealVisXL_V4.0"

pipe = StableDiffusionXLPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    variant="fp16",
    use_safetensors=True
)
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config, use_karras_sigmas=True)
pipe = pipe.to("cuda")

pipe.vae.enable_slicing()
pipe.vae.enable_tiling()

compel = Compel(
    tokenizer=[pipe.tokenizer, pipe.tokenizer_2],
    text_encoder=[pipe.text_encoder, pipe.text_encoder_2],
    returned_embeddings_type=ReturnedEmbeddingsType.PENULTIMATE_HIDDEN_STATES_NON_NORMALIZED,
    requires_pooled=[False, True],
    device="cuda"
)
print("✅ Model loaded and ready for dynamic generation.")

# ==========================================
# 2. GENERATION FUNCTION
# ==========================================
# Words that could make the model draw a person (matched as whole words,
# with optional plural / possessive: "grandma's", "kids", ...)
FORBIDDEN_WORDS = [
    "grandma", "grandmother", "grandpa", "grandfather", "granny", "grandparent",
    "mother", "mom", "mum", "father", "dad", "parent", "sister", "brother",
    "aunt", "uncle", "cousin", "family", "friend", "neighbor", "neighbour",
    "people", "person", "human", "child", "children", "kid", "baby", "boy", "girl",
    "man", "men", "woman", "women", "lady", "gentleman", "teacher", "student",
    "face", "portrait", "smile", "smiling", "laugh", "laughing", "laughter",
    "hand", "hands", "hug", "hugging", "wearing", "dress", "crowd", "figure",
]
FORBIDDEN_RE = re.compile(
    r"\b(" + "|".join(FORBIDDEN_WORDS) + r")(s|es|'s)?\b", re.IGNORECASE
)

def is_clean(text):
    return FORBIDDEN_RE.search(text) is None

def generate_memory_image(chapter_dict, output_filename):
    raw_location = chapter_dict.get('location') or 'a quiet space'
    raw_sensory = chapter_dict.get('sensory_details') or ['everyday items']

    # 1. Python filter: removes anything that could bring a person into the picture
    clean_sensory = [item for item in raw_sensory if is_clean(item)]
    sensory_items = ", ".join(clean_sensory) if clean_sensory else "everyday items"

    clean_loc_words = [w for w in raw_location.split() if is_clean(w)]
    safe_location = " ".join(clean_loc_words) if clean_loc_words else "quiet space"

    print(f"\n🎨 Generating illustration for: {output_filename}")
    print(f"🌍 Safe Location: '{safe_location}'")
    print(f"📖 Safe Objects: '{sensory_items}'")

    # 2. Positive prompt: describes an EMPTY scene, never mentions people
    #    (image models can't understand "no people" - mentioning them adds them)
    matched_style_prompt = (
        "A still life 35mm film photograph of a deserted, quiet, uninhabited space. "
        f"Foreground focus on: {sensory_items}. "
        f"Setting: inside an authentic, lived-in {safe_location}, completely empty and still. "
        "The background reveals the true, natural architecture of this space, featuring tools, reference materials, "
        f"and equipment belonging natively to a {safe_location}. "
        "The environment feels naturally in-use and functional, with authentic textures appropriate for this specific place. "
        "Lighting: Intense, natural golden-hour daytime sunlight forcefully streaming into the space, casting "
        "long organic shadows, and creating thick, dreamy, warm volumetric sunbeams with a light atmospheric haze "
        "that catches drifting dust motes. Natural lens flare and soft bloom on reflective surfaces. "
        "Camera: handheld snapshot, f/2.4, subtle camera grain, natural camera imperfections, shallow depth "
        "of field with beautiful bokeh, capturing rich, warm, earthy tones (oranges, gold, amber, and deep shadows). "
        "Nobody is present, an empty scene of objects and places only."
    )

    # 3. Negative prompt: strong, weighted block of everything human
    negative_prompt = (
        "(person, people, human, man, woman, girl, boy, child, baby, face, portrait, selfie, "
        "smile, eyes, hands, body, crowd, figure, silhouette, model, posing)1.6, "
        "hoarder, extreme clutter, piles of paper, garbage, trash, chaotic mess, ruined, destroyed, dirty, "
        "sterile, clinical, modern corporate office, flat white walls, sleek, brand new equipment, "
        "tidy, perfectly organized, corporate luxury, smooth digital render, CGI, glossy, 3D, plastic, "
        "blue lighting, cold lighting, fluorescent lights, overexposed, flat lighting"
    )

    conditioning, pooled = compel(matched_style_prompt)
    negative_conditioning, negative_pooled = compel(negative_prompt)

    gc.collect()
    torch.cuda.empty_cache()

    image = pipe(
        prompt_embeds=conditioning,
        pooled_prompt_embeds=pooled,
        negative_prompt_embeds=negative_conditioning,
        negative_pooled_prompt_embeds=negative_pooled,
        num_inference_steps=30,
        guidance_scale=7.5,
    ).images[0]

    image.save(output_filename)
    print(f"✅ Saved to {output_filename}")
    display(image)

    return output_filename

# ==========================================
# 3. THE FASTAPI SERVER
# ==========================================
class IncomingMemoir(BaseModel):
    title: str
    life_era: str
    estimated_year: Optional[int]
    location: Optional[str]
    key_people: List[str]
    sensory_details: List[str]
    core_wisdom: str
    narrative_text: str
    illustration_prompt: str

app = FastAPI()

@app.post("/generate-image")
def receive_and_paint(memoir: IncomingMemoir):
    print("📥 Received memory from Qwen! Generating image now...")
    output_filename = "qwen_memory_output.png"
    generate_memory_image(memoir.model_dump(), output_filename)
    return FileResponse(output_filename, media_type="image/png")

# --- NGROK ---
ngrok.kill()
ngrok.set_auth_token("3KUpJlS5dxmLn7VcjGEFWb25BaR_6VFH4ftyAWYFEZ1bqBkKZ")

public_url = ngrok.connect(8000).public_url
print(f"🚀 YOUR COLAB API URL IS: {public_url}/generate-image")

nest_asyncio.apply()
config = uvicorn.Config(app, host="0.0.0.0", port=8000)
server = uvicorn.Server(config)
import asyncio
asyncio.run(server.serve())
```
